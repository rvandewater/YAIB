```python
from numpy import genfromtxt
from icu_benchmarks.models.alarm_metrics import silence_positives, fill_gaps, convert_to_alarm
import numpy as np

predictions_no_threshold = genfromtxt(r"C:\Users\Robin\Downloads\pred_indicators_only_complications.csv", delimiter=",")
predictions = predictions_no_threshold.copy()
# Define the threshold
threshold = 0.5

# Round predictions to 0 and 1 according to the threshold
predictions = np.where(predictions >= threshold, 1, 0)

# Column index for grouping
group_column_index = 0

# Get unique group values
unique_groups = np.unique(predictions[:, group_column_index])
# Dictionary to store grouped arrays
grouped_arrays = {}

unique_groups = np.unique(predictions[:, group_column_index])

modified_arrays = []
# print(predictions)
# Group arrays based on unique values in the group column
for group in unique_groups:
    group_array = predictions[predictions[:, group_column_index] == group]
    # print(f"Group {group} before modification:\n{group_array}\n")
    ground_truth = group_array[:, 2]
    predictions = group_array[:, 4]
    group_array[:, 4] = silence_positives(ground_truth, predictions)
    group_array[:, 4], group_array[:, 2] = fill_gaps(predictions, ground_truth)
    # print(f"Group {group} after modification:\n{group_array}\n")
    modified_arrays.append(group_array)

# Combine the modified arrays back into a single array
combined_array = np.vstack(modified_arrays)
# combined_array = np.delete(combined_array, 3, axis=1)

# print("Combined array:")
# print(combined_array)
#
# print(f"ROC AUC: {roc_auc_score(predictions[:, 2], predictions[:, 4])}")
# print(f"ROC AUC: {roc_auc_score(combined_array[:, 2], combined_array[:, 4])}")
# print(f"Average precision: {average_precision_score(predictions[:, 2], predictions[:, 4])}")
# print(f"Average precision: {average_precision_score(combined_array[:, 2], combined_array[:, 4])}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import PrecisionRecallDisplay, RocCurveDisplay, average_precision_score, roc_auc_score

# Assuming predictions, combined_array, and predictions_no_threshold are already defined
# predictions[:, 2] is the ground truth
# predictions[:, 4] is the original predictions
# combined_array[:, 4] is the modified predictions
# predictions_no_threshold[:, 4] is the no_threshold predictions

# Calculate metrics
roc_auc_original = roc_auc_score(predictions[:, 2], predictions[:, 4])
roc_auc_modified = roc_auc_score(combined_array[:, 2], combined_array[:, 4])
roc_auc_no_threshold = roc_auc_score(predictions[:, 2], predictions_no_threshold[:, 4])
average_precision_original = average_precision_score(predictions[:, 2], predictions[:, 4])
average_precision_modified = average_precision_score(combined_array[:, 2], combined_array[:, 4])
average_precision_no_threshold = average_precision_score(predictions[:, 2], predictions_no_threshold[:, 4])

# Plot ROC curves
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))

RocCurveDisplay.from_predictions(predictions[:, 2], predictions[:, 4], ax=ax1, name="Original")
RocCurveDisplay.from_predictions(combined_array[:, 2], combined_array[:, 4], ax=ax1, name="Modified")
RocCurveDisplay.from_predictions(predictions[:, 2], predictions_no_threshold[:, 4], ax=ax1, name="No Threshold")
ax1.set_title(
    f"ROC Curve\nOriginal AUC: {roc_auc_original:.2f}, Modified AUC: {roc_auc_modified:.2f}, No Threshold AUC: {roc_auc_no_threshold:.2f}"
)

# Plot PRC curves
PrecisionRecallDisplay.from_predictions(predictions[:, 2], predictions[:, 4], ax=ax2, name="Original")
PrecisionRecallDisplay.from_predictions(combined_array[:, 2], combined_array[:, 4], ax=ax2, name="Modified")
PrecisionRecallDisplay.from_predictions(predictions[:, 2], predictions_no_threshold[:, 4], ax=ax2, name="No Threshold")
ax2.set_title(
    f"Precision-Recall Curve\nOriginal AP: {average_precision_original:.2f}, Modified AP: {average_precision_modified:.2f}, No Threshold AP: {average_precision_no_threshold:.2f}"
)

plt.tight_layout()
plt.show()
```

```python
pred_len = 24
label_horizon = 6
ground_truth = np.zeros(pred_len, dtype=int)
ground_truth[-label_horizon:] = 1
print(f"Ground truth: {ground_truth}")

# Simulate the model predictions (for example, using random values between 0 and 1)
predictions = np.random.rand(pred_len)
threshold = 0.5
predictions = np.where(predictions >= threshold, 1, 0)
print(f"Unaltered predictions: {predictions}")

# grace_horizon > label_horizon that we want to consider (i.e., the number of hours before the actual complication when the model is allowed but not required to predict 1)
grace_horizon = label_horizon * 2
# For how long should the alarm be silenced?
silencing_length = label_horizon
predictions = silence_positives(ground_truth, predictions, grace_horizon=grace_horizon, silencing_length=silencing_length)
print(f"Silenced predictions: {predictions}")

# Fill gaps in the predictions: we also want to consider early predictions with the grace_horizon.
predictions = fill_gaps(predictions)
print(f"Filled + silenced predictions: {predictions}")
```
